```python
from pyspark.sql import functions as F

SOURCE_PATH = "/Volumes/workspace/default/fraud_data/creditcard.csv"

raw_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(SOURCE_PATH)
)

print("Dataset rows:", raw_df.count())
print("Dataset columns:", len(raw_df.columns))
display(raw_df.limit(10))
```

```python
print("Schema:")
raw_df.printSchema()

print("Class distribution:")
display(
    raw_df.groupBy("Class")
    .count()
    .orderBy("Class")
)

print("Missing values by columns:")
missing_countas = raw_df.select([
    F.sum(
        F.when(F.col(c).isNull(),1).otherwise(0)
    ).alias(c)
    for c in raw_df.columns
])
```

```python
catalog_name = spark.sql(
    "SELECT current_catalog()"
).first()[0]

schema_name = spark.sql(
    "SELECT current_schema()"
).first()[0]

RAW_TABLE = f"{catalog_name}.{schema_name}.fraud_transactions_raw"
(
    raw_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(RAW_TABLE)
)

print("Saved Delta table:", RAW_TABLE)
```
